```python
import pandas as pd

# Read the files
matched_df = pd.read_excel('kroencke_taxa_matched.xlsx')
kroncke_df = pd.read_excel('Kroncke_1994_1998.xlsx', sheet_name='Biota')

# Merge the dataframes using ScientificName as key
kroncke_df = kroncke_df.merge(
    matched_df[['ScientificName', 'AphiaID_accepted', 'ScientificName_accepted']], 
    on='ScientificName', 
    how='left'
)

kroncke_df.to_excel('kroncke_wormed.xlsx', index=False)
# Display the updated dataframe
kroncke_df
```
