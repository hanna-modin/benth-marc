```python
import os
import pandas as pd
from pathlib import Path

os.chdir ('C:\\Users\\hamod4691\\Code\\arctic-benthic-biodiversity')
```

```python
gbif_path = os.path.join("GBIF/datasets_GBIF.parquet")
obis_path = os.path.join("OBIS/datasets_OBIS.parquet")

gbif_df = pd.read_parquet(gbif_path)
obis_df = pd.read_parquet(obis_path)

# Align OBIS column names with GBIF
obis_df = obis_df.rename(columns={
    "id": "alternateIdentifier",
    "published": "pubDate",
    "intellectualrights": "intellectualRights"
})

# Collapse embedded newlines and repeated whitespace in text fields
for frame in (gbif_df, obis_df):
    text_columns = frame.select_dtypes(include=["object", "string"]).columns
    for column in text_columns:
        frame[column] = (
            frame[column]
            .astype("string")
            .str.replace(r"\s+", " ", regex=True)
            .str.strip()
        )

# Mark the source dataset for each row
gbif_df["source"] = "GBIF"
obis_df["source"] = "OBIS"

# Merge by stacking rows, keeping all columns from both files
merged_df = pd.concat([gbif_df, obis_df], ignore_index=True, sort=False)

output_path = os.path.join(os.getcwd(), "Datasets/datasets_GBIF_OBIS.xlsx")
merged_df.to_excel(output_path, index=False)

print(f"Saved: {output_path}")
```

```python
manual_path = os.path.join("Datasets", "datasets_GBIF_OBIS_manual.xlsx")
target_path = os.path.join("Datasets", "datasets_GBIF_OBIS.xlsx")

manual_df = pd.read_excel(manual_path)
target_df = pd.read_excel(target_path)

keep_map = manual_df.set_index("alternateIdentifier")["Keep"]
target_df["Keep"] = target_df["alternateIdentifier"].map(keep_map)

target_df.to_excel(target_path, index=False)
print(f"Saved: {target_path}")
```
