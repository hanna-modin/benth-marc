```python
from pygbif import species
import pandas as pd

phylum_keys = [53, 105, 50, 75, 5967457, 74, 5963150, 63, 110, 5959089, 14]
class_keys = [137, 276, 346, 356, 256, 8365233, 354, 8066236, 309, 206]
order_keys = [705, 1415, 12204028, 1426, 1242, 1500, 1078, 5958700]
family_keys = [7325, 6960, 4531, 6191, 5379, 8691, 8146, 3363, 9676, 4529, 8888, 8895, 9272]
genus_keys = [8309288, 2268054, 2228475, 2268684, 2207284, 4283770, 2265855, 2202982]

def get_taxa_names(keys, rank):
    results = []
    for key in keys:
        try:
            result = species.name_usage(key=key)
            results.append({'key': key, 'rank': rank, 'name': result.get('scientificName', 'N/A')})
        except Exception as e:
            results.append({'key': key, 'rank': rank, 'name': 'Error'})
    return results

all_taxa = []
all_taxa.extend(get_taxa_names(phylum_keys, 'phylum'))
all_taxa.extend(get_taxa_names(class_keys, 'class'))
all_taxa.extend(get_taxa_names(order_keys, 'order'))
all_taxa.extend(get_taxa_names(family_keys, 'family'))
all_taxa.extend(get_taxa_names(genus_keys, 'genus'))

gbif_df = pd.DataFrame(all_taxa)
gbif_df.to_csv('gbif_obligate_benthos.csv', index=False)
print(gbif_df)
```
