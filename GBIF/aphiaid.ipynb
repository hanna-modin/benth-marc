```python
import pandas as pd
import os

os.chdir ('C:\\Users\\hamod4691\\Code\\arctic-benthic-biodiversity\\GBIF')

# Load the parquet file
df = pd.read_parquet('records_keep_notCB.parquet')

# Get unique entries for both columns
unique_taxonKey = df['taxonKey'].unique()
unique_acceptedScientificName = df['acceptedScientificName'].unique()

print("Unique taxonKey entries:")
print(unique_taxonKey)
print(f"\nTotal unique taxonKey: {len(unique_taxonKey)}")

print("\n" + "="*50)
print("Unique acceptedScientificName entries:")
print(unique_acceptedScientificName)
print(f"\nTotal unique acceptedScientificName: {len(unique_acceptedScientificName)}")

# Column coverage analysis
print("\n" + "="*50)
print("Column Coverage Analysis:")
print(f"Total rows: {len(df)}")
print(f"\ntaxonKey:")
print(f"  Non-null values: {df['taxonKey'].notna().sum()}")
print(f"  Coverage: {df['taxonKey'].notna().sum() / len(df) * 100:.2f}%")
print(f"  Unique values: {df['taxonKey'].nunique()}")

print(f"\nacceptedScientificName:")
print(f"  Non-null values: {df['acceptedScientificName'].notna().sum()}")
print(f"  Coverage: {df['acceptedScientificName'].notna().sum() / len(df) * 100:.2f}%")
print(f"  Unique values: {df['acceptedScientificName'].nunique()}")

# Show examples of multiple taxonKeys for same name
print("\n" + "="*50)
print("Examples of multiple taxonKeys per acceptedScientificName:")
name_counts = df.groupby('acceptedScientificName')['taxonKey'].nunique().sort_values(ascending=False)
print(name_counts.head(10))

# Show specific taxonKeys for Cyclopecten hoskynsi
print("\n" + "="*50)
print("TaxonKeys for 'Cyclopecten hoskynsi (Forbes, 1844)':")
cyclopecten_keys = df[df['acceptedScientificName'] == 'Cyclopecten hoskynsi (Forbes, 1844)']['taxonKey'].unique()
print(cyclopecten_keys)
print(f"Total: {len(cyclopecten_keys)}")

# Analyze how common multiple taxonKeys per name is
print("\n" + "="*50)
print("Distribution of taxonKeys per acceptedScientificName:")
keys_per_name = df.groupby('acceptedScientificName')['taxonKey'].nunique()
print(f"\nNames with exactly 1 taxonKey: {(keys_per_name == 1).sum()} ({(keys_per_name == 1).sum() / len(keys_per_name) * 100:.2f}%)")
print(f"Names with 2+ taxonKeys: {(keys_per_name > 1).sum()} ({(keys_per_name > 1).sum() / len(keys_per_name) * 100:.2f}%)")
print(f"Names with 3+ taxonKeys: {(keys_per_name >= 3).sum()} ({(keys_per_name >= 3).sum() / len(keys_per_name) * 100:.2f}%)")
print(f"Names with 5+ taxonKeys: {(keys_per_name >= 5).sum()} ({(keys_per_name >= 5).sum() / len(keys_per_name) * 100:.2f}%)")
print(f"\nAverage taxonKeys per name: {keys_per_name.mean():.2f}")
print(f"Median taxonKeys per name: {keys_per_name.median():.0f}")
print(f"Max taxonKeys for a single name: {keys_per_name.max()}")

# Create summary statistics dataframe
print("\n" + "="*50)
print("Summary Statistics Table:")
summary_stats = pd.DataFrame({
    'Metric': [
        'Total Rows',
        'Unique taxonKeys',
        'Unique Scientific Names',
        'TaxonKey Coverage (%)',
        'Name Coverage (%)',
        'Names with 1 taxonKey',
        'Names with 2+ taxonKeys',
        'Names with 3+ taxonKeys',
        'Names with 5+ taxonKeys',
        'Avg taxonKeys/name',
        'Median taxonKeys/name',
        'Max taxonKeys/name'
    ],
    'Value': [
        len(df),
        df['taxonKey'].nunique(),
        df['acceptedScientificName'].nunique(),
        f"{df['taxonKey'].notna().sum() / len(df) * 100:.2f}%",
        f"{df['acceptedScientificName'].notna().sum() / len(df) * 100:.2f}%",
        f"{(keys_per_name == 1).sum()} ({(keys_per_name == 1).sum() / len(keys_per_name) * 100:.2f}%)",
        f"{(keys_per_name > 1).sum()} ({(keys_per_name > 1).sum() / len(keys_per_name) * 100:.2f}%)",
        f"{(keys_per_name >= 3).sum()} ({(keys_per_name >= 3).sum() / len(keys_per_name) * 100:.2f}%)",
        f"{(keys_per_name >= 5).sum()} ({(keys_per_name >= 5).sum() / len(keys_per_name) * 100:.2f}%)",
        f"{keys_per_name.mean():.2f}",
        f"{keys_per_name.median():.0f}",
        keys_per_name.max()
    ]
})
print(summary_stats.to_string(index=False))

# Save both to one CSV file
unique_df = pd.DataFrame({
    'taxonKey': pd.Series(unique_taxonKey),
    'acceptedScientificName': pd.Series(unique_acceptedScientificName)
})
unique_df.to_csv('unique_records.csv', index=False)
print(f"\nSaved to unique_records.csv")
```

```python
# Create a dataframe with unique acceptedScientificName and their taxonRank
unique_rank_df = df[['acceptedScientificName', 'taxonRank']].drop_duplicates().reset_index(drop=True)

# Save to CSV
unique_rank_df.to_csv('unique_names_with_ranks.csv', index=False)
print(f"Saved {len(unique_rank_df)} unique entries to unique_names_with_ranks.csv")
print(unique_rank_df.head(10))
```

```python
# Create a function to clean the names based on taxon rank
def clean_name(row):
    name = row['acceptedScientificName']
    rank = row['taxonRank']
    
    if pd.isna(name):
        return name, None
    
    words = name.split()
    
    if rank == 'GENUS':
        # Keep only the first word
        cleaned = words[0]
        removed = ' '.join(words[1:]) if len(words) > 1 else None
    elif rank == 'SPECIES':
        # Keep only the first two words
        cleaned = ' '.join(words[:2]) if len(words) >= 2 else name
        removed = ' '.join(words[2:]) if len(words) > 2 else None
    else:
        # Keep the full name
        cleaned = name
        removed = None
    
    return cleaned, removed

# Apply the cleaning function
unique_rank_df[['cleaned_name', 'removed_part']] = unique_rank_df.apply(clean_name, axis=1, result_type='expand')

# Create dataframes for saving
unique_names_cleaned = unique_rank_df[['acceptedScientificName', 'taxonRank', 'cleaned_name']].copy()
unique_names_cleaned.rename(columns={'acceptedScientificName': 'original_name'}, inplace=True)

unique_names_removed = unique_rank_df[['acceptedScientificName', 'taxonRank', 'removed_part']].dropna(subset=['removed_part']).copy()
unique_names_removed.rename(columns={'acceptedScientificName': 'original_name'}, inplace=True)

# Save to CSV
unique_names_cleaned.to_csv('unique_names_cleaned.csv', index=False)
unique_names_removed.to_csv('unique_names_removed.csv', index=False)

print(f"Saved {len(unique_names_cleaned)} cleaned names to unique_names_cleaned.csv")
print(f"Saved {len(unique_names_removed)} removed parts to unique_names_removed.csv")
print("\nSample of cleaned names:")
print(unique_names_cleaned.head(15))
print("\nSample of removed parts:")
print(unique_names_removed.head(10))
```

```python
# Create a dataframe with only the cleaned_name column
cleaned_names_only = unique_names_cleaned[['cleaned_name']].copy()

# Save to CSV
cleaned_names_only.to_csv('unique_names_worms.csv', index=False)

print(f"Saved {len(cleaned_names_only)} cleaned names to unique_names_worms.csv")
print("\nSample of cleaned names:")
print(cleaned_names_only.head(15))
```
