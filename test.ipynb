```python
import pandas as pd

# Load files
base = pd.read_excel("taxonomic_filter.xlsx")
env = pd.read_excel("taxonomic_filter_with_environment.xlsx")[["AphiaID", "worms_environment"]].drop_duplicates(subset=["AphiaID"])
extinct = pd.read_csv("taxonomic_filter_extinct_results.csv")[["AphiaID", "worms_isExtinct"]].drop_duplicates(subset=["AphiaID"])

# Make sure keys match exactly before merging
base["AphiaID"] = pd.to_numeric(base["AphiaID"], errors="coerce").astype("Int64")
env["AphiaID"] = pd.to_numeric(env["AphiaID"], errors="coerce").astype("Int64")
extinct["AphiaID"] = pd.to_numeric(extinct["AphiaID"], errors="coerce").astype("Int64")

# Merge by AphiaID
result = (
    base.merge(env, on="AphiaID", how="left")
        .merge(extinct, on="AphiaID", how="left")
)
result = result.drop_duplicates().reset_index(drop=True)

# Save updated file
result.to_excel("taxonomic_filter.xlsx", index=False)
```

```python
# Remove fully duplicated rows from the merged dataframe and overwrite the Excel file
before_rows = len(result)
result = result.drop_duplicates().reset_index(drop=True)
after_rows = len(result)

result.to_excel("taxonomic_filter_updated.xlsx", index=False)

print(f"Removed {before_rows - after_rows} duplicate rows.")
print(f"Final row count: {after_rows}")
```

```python
# Remove fully duplicated rows from the original taxonomic filter file and overwrite it
before = len(base)
base = base.drop_duplicates().reset_index(drop=True)
after = len(base)

base.to_excel("taxonomic_filter_manual.xlsx", index=False)

print(f"Removed {before - after} duplicate rows from taxonomic_filter_manual.xlsx.")
print(f"Final row count: {after}")
```
